```python
# Cell 1: Markdown
# Flight Delay Data Exploration and Model Building
#
# This notebook loads, cleanses, and explores the FAA 2013 US flights dataset.
# It builds a model to predict the chance of a flight being delayed by more than 15 minutes for a given day and airport,
# and exports a CSV of all airports and their IDs.

# Cell 2: Import libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
import joblib

# Cell 3: Load the data
df = pd.read_csv('data/flights.csv')
df.head()

# Cell 4: Check for null values
df.isnull().sum()

# Cell 5: Replace null values with zero
df_filled = df.fillna(0)
df_filled.isnull().sum()

# Cell 6: Preview cleaned data
df_filled.head()

# Cell 7: Prepare features and target for modeling
features = df_filled[['DayOfWeek', 'OriginAirportID']]
target = df_filled['DepDel15']

# Ensure target is binary (0 or 1)
target = target.apply(lambda x: 1 if x == 1 else 0)

# Encode categorical variables
features_encoded = pd.get_dummies(features, columns=['OriginAirportID'])

# Cell 8: Split data
X_train, X_test, y_train, y_test = train_test_split(features_encoded, target, test_size=0.2, random_state=42)

# Cell 9: Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Cell 10: Evaluate model
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))

# Cell 11: Save the model
joblib.dump(model, 'flight_delay_model.joblib')

# Cell 12: Export airport IDs and names
airports = df_filled[['OriginAirportID', 'OriginAirportName']].drop_duplicates()
airports.to_csv('airport_ids.csv', index=False)
```
